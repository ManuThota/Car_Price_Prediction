# Encoding Catogorical Data

We use ONE HOT ENCODING to Encode Catogorical Data

```python
data = pd.get_dummies(data, columns=["Fuel_Type", "Seller_Type", "Transmission"],drop_first=True)
```

# Splitting Data

We don't need "Car_Name" column, so removing it.

```python
data = data.drop(['Car_Name'], axis=1)
```

Splitting the Data and target

```python
x = data.drop(['Selling_Price'], axis=1)
y = data['Selling_Price']
```

```python
x.head()
```

```python
y.head()
```

Splitting Data and Target into Training and Testing data

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.1, random_state = 2)
```

```python
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

# Machine Learning

# 1. Linear Regression

```python
from sklearn.linear_model import LinearRegression

lin_model = LinearRegression()
```

Loading the Traing Data into the model

```python
lin_model.fit(x_train, y_train)
```

 Report on Traning

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

lin_train_pred = lin_model.predict(x_train)

print(f"R-squared : {r2_score(y_train, lin_train_pred):.3f}")
print(f"Mean Absolute Error : {mean_absolute_error(y_train, lin_train_pred):.3f}")
print(f"Mean Squared Error : {mean_squared_error(y_train, lin_train_pred):.3f}")
```

Report on Testing

```python
lin_test_pred = lin_model.predict(x_test)

print(f"R-squared : {r2_score(y_test, lin_test_pred):.3f}")
print(f"Mean Absolute Error : {mean_absolute_error(y_test, lin_test_pred):.3f}")
print(f"Mean Squared Error : {mean_squared_error(y_test, lin_test_pred):.3f}")
```

```python
import numpy as np

print(f"RMSE (Train): {np.sqrt(mean_squared_error(y_train, lin_train_pred)):.3f}")
print(f"RMSE (Test) : {np.sqrt(mean_squared_error(y_test, lin_test_pred)):.3f}")
```

Residual Analysis (Quick Check)

```python
lin_train_residuals = y_train - lin_train_pred
lin_test_residuals = y_test - lin_test_pred

print("Mean Train Residual:", np.mean(lin_train_residuals))
print("Mean Test Residual :", np.mean(lin_test_residuals))
```

**Conclusion on Linear Regression**

• Linear Regression shows good baseline performance with an R² of 0.88 (train) and 0.86 (test).

• Error metrics (MAE ≈ 1.07, RMSE ≈ 1.31 on test data) indicate moderate prediction accuracy.

• Training and test results are reasonably close, suggesting low overfitting.

• Residual means are close to zero, indicating the model is largely unbiased.

# 3. Random Forest Regression


```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
```

Loading Training Data into the model

```python
rf_model.fit(x_train, y_train)
```

Report on Training

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

rf_train_pred = rf_model.predict(x_train)

print(f"R-squared : {r2_score(y_train, rf_train_pred):.3f}")
print(f"MAE       : {mean_absolute_error(y_train, rf_train_pred):.3f}")
print(f"MSE       : {mean_squared_error(y_train, rf_train_pred):.3f}")
```

Report on Testing

```python
rf_test_pred = rf_model.predict(x_test)

print(f"R-squared : {r2_score(y_test, rf_test_pred):.3f}")
print(f"MAE       : {mean_absolute_error(y_test, rf_test_pred):.3f}")
print(f"MSE       : {mean_squared_error(y_test, rf_test_pred):.3f}")
```

```python
import numpy as np

print(f"RMSE (Train): {np.sqrt(mean_squared_error(y_train, rf_train_pred)):.3f}")
print(f"RMSE (Test) : {np.sqrt(mean_squared_error(y_test, rf_test_pred)):.3f}")
```

Residual Analysis Quick Check

```python
rf_train_residuals = y_train - rf_train_pred
rf_test_residuals = y_test - rf_test_pred

print("Mean Train Residual:", np.mean(rf_train_residuals))
print("Mean Test Residual :", np.mean(rf_test_residuals))
```

**Conclusion on Random Forest Regressor**

• The model shows excellent performance with very high R² scores (0.98 on training and 0.98 on test data), indicating strong predictive power.

• Very low MAE, MSE, and RMSE values demonstrate highly accurate car price predictions.

• Training and test results are very close, suggesting minimal overfitting and strong generalization.

• Mean residuals are close to zero, indicating the model is largely unbiased in its predictions.

# Final Conclusion: Which model to select?

Random Forest was selected as the final model because

• Superior Precision: It reduces the average error (MAE) by approximately 69% compared to the linear models.

• High Explained Variance: It captures 98% of the data's behavior, whereas the linear models miss about 13% of the underlying patterns.

• Non-Linearity: The massive improvement in $R^2$ suggests that the relationship between your features and the target variable is likely non-linear, which the Random Forest handles naturally, while the linear models struggle.

• Stability: Despite being a complex ensemble model, the test results remain incredibly close to the training results, proving it is a reliable and stable predictor.
